```python
import pandas as pd
import os
from IPython.display import Audio, display
import matplotlib.pyplot as plt
import random
    
# Load splits
train_df = pd.read_csv('../data/splits/train.csv')
val_df = pd.read_csv('../data/splits/val.csv')
test_df = pd.read_csv('../data/splits/test.csv')
    
print(f'Train: {len(train_df)}, Val: {len(val_df)}, Test: {len(test_df)}')
print('\nClass distribution (Train):')
print(train_df['label'].value_counts(normalize=True))
```

```python
# Function to play random samples,
# def play_random_samples(df, n=3):
#     samples = df.sample(n)
#     for row in samples.iterrows():
#         label = 'Positive' if row['label'] == 1 else 'Negative',
#         print(f"{label} - {os.path.basename(row['file_path'])}")
#         display(Audio(row['file_path'])),
    
# print('Random Positive Samples:')
# play_random_samples(train_df[train_df['label'] == 1], 3)
    
# print('\Random Negative Samples:')
# play_random_samples(train_df[train_df['label'] == 0], 3)

def play_random_samples(df, n=3):
    samples = df.sample(n)
    for _, row in samples.iterrows():   # unpack tuple
        label = 'Positive' if row['label'] == 1 else 'Negative'
        print(f"{label} - {os.path.basename(row['file_path'])}")
        display(Audio(row['file_path']))

print('Random Positive Samples:')
play_random_samples(train_df[train_df['label'] == 1], 3)

print('Random Negative Samples:')
play_random_samples(train_df[train_df['label'] == 0], 3)
```
